```python
import sys
import os
sys.path.insert(0, os.path.abspath('..'))

import pickle
import pandas as pd
from pprint import pprint
```

## Specify Simulation Directory

Set the path to your simulation directory here:

```python
# Example simulation directory - replace with your own
sim_dir = "/home/drfrbc/Neural-Modeling/examples/example_simulations/sta_Complex_NexInhDen0.62_Np5000"

# Check if directory exists
if not os.path.exists(sim_dir):
    print(f"❌ Directory not found: {sim_dir}")
else:
    print(f"✓ Directory found: {sim_dir}")
```

## Load Parameters

Load the parameters from `parameters.pickle`:

```python
params_file = os.path.join(sim_dir, "parameters.pickle")

if not os.path.exists(params_file):
    print(f"❌ Parameters file not found: {params_file}")
else:
    with open(params_file, 'rb') as f:
        parameters = pickle.load(f)
    print("✓ Parameters loaded successfully")
    print(f"Parameter object type: {type(parameters)}")
```

## Display All Parameters

Show all parameter attributes:

```python
# Get all attributes (excluding private/magic methods)
param_names = [attr for attr in dir(parameters) if not attr.startswith('_')]

print(f"Total parameters: {len(param_names)}\n")
print("Available parameters:")
print("-" * 50)
for name in sorted(param_names):
    print(f"  • {name}")
```

## Display Key Simulation Parameters

Show commonly used parameters:

```python
# Display key parameters in organized groups
print("="*60)
print("SIMULATION SETTINGS")
print("="*60)

# Time parameters
print("\n📊 Time Parameters:")
print(f"  Duration (h_tstop):     {getattr(parameters, 'h_tstop', 'N/A')} ms")
print(f"  Time step (h_dt):       {getattr(parameters, 'h_dt', 'N/A')} ms")
print(f"  Save every:             {getattr(parameters, 'save_every_ms', 'N/A')} ms")

# Cell parameters
print("\n🧬 Cell Parameters:")
print(f"  Skeleton cell type:     {getattr(parameters, 'skeleton_cell_type', 'N/A')}")
print(f"  Simulation type:        {getattr(parameters, 'sim_type', 'N/A')}")

# Synapse parameters
print("\n⚡ Synapse Parameters:")
print(f"  Excitatory mod:         {getattr(parameters, 'exc_syn_mod', 'N/A')}")
print(f"  Inhibitory mod:         {getattr(parameters, 'inh_syn_mod', 'N/A')}")
print(f"  Total synapses (Np):    {getattr(parameters, 'Np', 'N/A')}")

# Recording parameters
print("\n📹 Recording Parameters:")
print(f"  Record all channels:    {getattr(parameters, 'record_all_channels', 'N/A')}")
print(f"  Record all synapses:    {getattr(parameters, 'record_all_synapses', 'N/A')}")
print(f"  Record voltage:         {getattr(parameters, 'record_v', 'N/A')}")
```

## Display Clustering Configuration

Show excitatory and inhibitory clustering settings:

```python
print("="*60)
print("CLUSTERING CONFIGURATION")
print("="*60)

# Excitatory clustering
exc_clustering = getattr(parameters, 'exc_clustering', None)
if exc_clustering:
    print("\n✓ Excitatory Clustering:")
    for input_source, config in exc_clustering.items():
        print(f"\n  {input_source}:")
        if 'functional_groups' in config:
            print(f"    • Number of FGs: {len(config['functional_groups'])}")
            if config['functional_groups']:
                fg = config['functional_groups'][0]
                print(f"    • FG radius (example): {fg.get('radius', 'N/A'):.2f} μm")
                pcs = fg.get('presynaptic_cells', [])
                if isinstance(pcs, list):
                    print(f"    • PCs per FG (example): {len(pcs)}")
                elif isinstance(pcs, dict):
                    print(f"    • PC mode: {pcs.get('mode', 'unknown')}")
else:
    print("\n❌ No excitatory clustering configuration found")

# Inhibitory clustering
inh_clustering = getattr(parameters, 'inh_clustering', None)
if inh_clustering:
    print("\n✓ Inhibitory Clustering:")
    for input_source, config in inh_clustering.items():
        print(f"\n  {input_source}:")
        if 'functional_groups' in config:
            print(f"    • Number of FGs: {len(config['functional_groups'])}")
else:
    print("\n❌ No inhibitory clustering configuration found")
```

## Display Synapse Properties

Show excitatory and inhibitory synapse properties by input source:

```python
print("="*60)
print("SYNAPSE PROPERTIES")
print("="*60)

# Excitatory synapse properties
exc_syn_props = getattr(parameters, 'exc_syn_properties', None)
if exc_syn_props:
    print("\n✓ Excitatory Synapse Properties:")
    for input_source, props in exc_syn_props.items():
        print(f"\n  {input_source}:")
        print(f"    • Syn density:     {props.get('syn_density', 'N/A')}")
        print(f"    • gmax:            {props.get('gmax', 'N/A')}")
        print(f"    • Mean FR:         {props.get('mean_FR', 'N/A')} Hz")
        print(f"    • Std FR:          {props.get('std_FR', 'N/A')} Hz")
else:
    print("\n❌ No excitatory synapse properties found")

# Inhibitory synapse properties
inh_syn_props = getattr(parameters, 'inh_syn_properties', None)
if inh_syn_props:
    print("\n✓ Inhibitory Synapse Properties:")
    for input_source, props in inh_syn_props.items():
        print(f"\n  {input_source}:")
        print(f"    • Syn density:     {props.get('syn_density', 'N/A')}")
        print(f"    • gmax:            {props.get('gmax', 'N/A')}")
        print(f"    • Mean FR:         {props.get('mean_FR', 'N/A')} Hz")
        print(f"    • Std FR:          {props.get('std_FR', 'N/A')} Hz")
else:
    print("\n❌ No inhibitory synapse properties found")
```

## Access Specific Parameter

Example of how to access and work with a specific parameter:

```python
# Example: Get simulation duration and calculate number of time steps
h_tstop = getattr(parameters, 'h_tstop', None)
h_dt = getattr(parameters, 'h_dt', None)

if h_tstop and h_dt:
    n_steps = int(h_tstop / h_dt)
    print(f"Simulation duration: {h_tstop} ms")
    print(f"Time step: {h_dt} ms")
    print(f"Total time steps: {n_steps:,}")
    print(f"Duration in seconds: {h_tstop/1000:.2f} s")
else:
    print("Could not calculate time steps")
```

## Export Parameters to DataFrame

Create a pandas DataFrame with key parameters for easy viewing:

```python
# Create a dictionary of key parameters
param_dict = {}

key_params = [
    'h_tstop', 'h_dt', 'skeleton_cell_type', 'sim_type',
    'exc_syn_mod', 'inh_syn_mod', 'Np',
    'record_all_channels', 'record_all_synapses', 'record_v',
    'save_every_ms'
]

for param in key_params:
    value = getattr(parameters, param, 'N/A')
    param_dict[param] = [value]

# Create DataFrame
params_df = pd.DataFrame(param_dict).T
params_df.columns = ['Value']
params_df.index.name = 'Parameter'

print("\nKey Parameters Summary:")
print("="*60)
display(params_df)
```
