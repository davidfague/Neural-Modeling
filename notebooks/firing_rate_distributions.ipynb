```python
def plot_distribution(function, func_params, n_samples, n_bins):
    """
    Plots the distribution of a given function with specified parameters.
    
    Parameters:
    - function: The function to plot.
    - func_params: Parameters for the function.
    - n_samples: Number of samples to generate.
    - n_bins: Number of bins for the histogram.
    """
    samples = function(**func_params, size=n_samples)
    plt.hist(samples, bins=n_bins, density=True, alpha=0.6, color='g')
    plt.title(f'Distribution of {function.__name__}')
    plt.xlabel('Value')
    plt.ylabel('Density')
    plt.grid()
    plt.show()

def compute_stats(function, func_params, n_samples):
    """
    Computes statistics of a given function with specified parameters.
    
    Parameters:
    - function: The function to compute statistics for.
    - func_params: Parameters for the function.
    - n_samples: Number of samples to generate.
    
    Returns:
    - mean: Mean of the samples.
    - std_dev: Standard deviation of the samples.
    """
    samples = function(**func_params, size=n_samples)
    mean = np.mean(samples)
    std_dev = np.std(samples)
    return mean, std_dev

def plot_and_compute_stats(function, func_params, n_samples=10000, n_bins=50):
    """
    Plots the distribution and computes statistics of a given function.
    
    Parameters:
    - function: The function to plot and compute statistics for.
    - func_params: Parameters for the function.
    - n_samples: Number of samples to generate.
    - n_bins: Number of bins for the histogram.
    
    Returns:
    - mean: Mean of the samples.
    - std_dev: Standard deviation of the samples.
    """
    plot_distribution(function, func_params, n_samples, n_bins)
    mean, std_dev = compute_stats(function, func_params, n_samples)
    print(f'Mean: {mean}, Standard Deviation: {std_dev}')
    return mean, std_dev

levy_params = {'alpha': 1.37, 'beta': -1.00, 'loc': 0.92, 'scale': 0.44}
plot_and_compute_stats(exp_levy_dist, levy_params, n_samples=10000, n_bins=50)
```
