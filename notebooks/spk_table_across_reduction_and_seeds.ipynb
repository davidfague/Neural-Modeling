```python
from Modules import analysis
import os

var_to_plot = ['Total_NMDA_Spikes_per_micron', 'Total_NA_Spikes_per_micron', 'Total_CA_Spikes_per_micron', 'soma_firing_rate']
# var_to_plot = ['Total_NMDA_Spikes', 'Total_NA_Spikes', 'Total_CA_Spikes', 'soma_firing_rate']

# Initialize lists to store results
soma_firing_rates = []
morphologies = []
seeds = []

# Compute the soma firing rate for each simulation directory
for sim_dir in spike_table.simulation_directory:
    sim_path = os.path.join(sim_directory, sim_dir)
    
    # Read soma spikes data
    spks = analysis.DataReader.read_data(sim_path, "soma_spikes")
    
    # Load simulation parameters to get h_tstop, morphology, and seed
    params = analysis.DataReader.load_parameters(sim_path)
    
    # Calculate soma firing rate (convert ms to seconds)
    firing_rate = len(spks[0][:]) / (params.h_tstop / 1000)
    soma_firing_rates.append(firing_rate)
    
    # Instead of rhythmic depths, record morphology and seed
    morphologies.append(params.morphology_name)
    seeds.append(params.numpy_random_state)  # Assumes params contains a seed attribute

# Add the computed columns to spike_table
spike_table['soma_firing_rate'] = soma_firing_rates
spike_table['morphology'] = morphologies
spike_table['seed'] = seeds

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# --- 1. Grouping by morphology, seed, and Segment_Type ---
grouped_stats = spike_table.groupby(
    ['morphology', 'seed', 'Segment_Type']
)[var_to_plot].agg(['mean', 'std'])

# Flatten the multi-index columns (e.g. Total_NMDA_Spikes_mean)
grouped_stats.columns = ['_'.join(col).strip() for col in grouped_stats.columns.values]
grouped_stats = grouped_stats.reset_index()

print(grouped_stats.head())

# --- 2. Plotting spike metrics as a function of morphology and seed ---

# List of spike types and corresponding titles
spike_types = var_to_plot[:-1]
titles = ['NMDA Spikes', 'NA Spikes', 'CA Spikes']

# Loop over each spike type and each segment type for heatmap plots (mean and std)
for spike_type, title in zip(spike_types, titles):
    for segment_type in grouped_stats['Segment_Type'].unique():
        # Subset data for the current segment type
        subset = grouped_stats[grouped_stats['Segment_Type'] == segment_type]
        
        # Pivot: rows = morphology, columns = seed
        pivot_mean = subset.pivot(index='morphology', 
                                  columns='seed', 
                                  values=f'{spike_type}_mean')
        pivot_std = subset.pivot(index='morphology', 
                                 columns='seed', 
                                 values=f'{spike_type}_std')
        
        # Create a figure with two subplots for mean and std
        fig, axes = plt.subplots(1, 2, figsize=(12, 5))
        
        sns.heatmap(pivot_mean, ax=axes[0], cmap="viridis", annot=True, fmt=".1e")
        axes[0].set_title(f'{title} Mean ({segment_type})')
        axes[0].set_xlabel('Seed')
        axes[0].set_ylabel('Morphology')
        
        sns.heatmap(pivot_std, ax=axes[1], cmap="magma", annot=True, fmt=".1e")
        axes[1].set_title(f'{title} Std ({segment_type})')
        axes[1].set_xlabel('Seed')
        axes[1].set_ylabel('Morphology')
        
        plt.tight_layout()
        plt.show()

# --- 3. Plotting Soma Firing Rate similarly ---
for segment_type in grouped_stats['Segment_Type'].unique():
    subset = grouped_stats[grouped_stats['Segment_Type'] == segment_type]
    pivot_mean = subset.pivot(index='morphology', columns='seed', values='soma_firing_rate_mean')
    pivot_std = subset.pivot(index='morphology', columns='seed', values='soma_firing_rate_std')
    
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    
    sns.heatmap(pivot_mean, ax=axes[0], cmap="viridis", annot=True, fmt=".1f")
    axes[0].set_title(f'Soma Firing Rate Mean ({segment_type})')
    axes[0].set_xlabel('Seed')
    axes[0].set_ylabel('Morphology')
    
    sns.heatmap(pivot_std, ax=axes[1], cmap="magma", annot=True, fmt=".1f")
    axes[1].set_title(f'Soma Firing Rate Std ({segment_type})')
    axes[1].set_xlabel('Seed')
    axes[1].set_ylabel('Morphology')
    
    plt.tight_layout()
    plt.show()
```
