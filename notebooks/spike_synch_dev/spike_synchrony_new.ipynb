# Spike Synchrony — Complete Analysis

This notebook performs **full-resolution** spike synchrony analysis across all hierarchical grouping levels:

**Within-group synchrony:**
1. Within `input_source` (e.g., distal_basal, tuft, oblique)
2. Within `input_source × functional_group`
3. Within `input_source × functional_group × presynaptic_cell`

**Between-group synchrony:**
4. Between `input_source` pairs
5. Between `functional_group` pairs (within same input_source)
6. Between `presynaptic_cell` pairs (within same fg & input_source)

Outputs are saved to `{sim_dir}/spike_synchrony/`:
- `spike_sync_summary.csv` - complete statistics table
- `sync_matrix_{sim_name}.png` - SPIKE-synchronization matrix heatmap
- `sync_comprehensive_analysis.png` - 6-panel visualization

```python
%matplotlib inline
import matplotlib.pyplot as plt
import pandas as pd
import os
import sys
sys.path.append('..')  # to access Modules/
import Modules.spike_synchrony as ss

# Simulation directory
SIM_DIR = '/home/drfrbc/Neural-Modeling/simulations/2025-10-22-19-58-0.16xTuftInh_0.05xBasalExc_0.05xTuftExc_1.5xTuftInh/allinh_delay_shift_4ms_Np5000'

print("Loading synapses (full dataset, no subsampling)...")
syn_df = ss.load_synapses([SIM_DIR])
print(f"Loaded {len(syn_df)} spike trains")

print("Preparing spike trains...")
syn_df, (t_start, t_end) = ss.prepare_spike_trains(syn_df, t_start=0.0, min_spikes=1)
print(f"Time window: [{t_start}, {t_end}]")

print("Assigning positions within simulation...")
syn_df = ss.assign_positions_within_sim(syn_df)

print("Computing SPIKE-synchronization matrices (this may take a few minutes)...")
sync_mats = ss.build_sync_matrices_per_sim(syn_df)
first_sim = next(iter(sync_mats.keys()))
print(f"Computed sync matrix: {sync_mats[first_sim].shape}")

print("\nComputing statistics across all 6 hierarchical levels...")
all_stats = ss.compute_all_stats(syn_df, sync_mats, drop_fg_negative=True)
print(f"Generated {len(all_stats)} summary rows")
print(f"Levels: {all_stats['level'].unique()}")

# Display sync matrix
fig, ax = plt.subplots(figsize=(8, 8))
ss.plot_sync_matrix(sync_mats, first_sim, ax=ax)
plt.show()

# Create comprehensive 6-panel visualization
print("\nCreating comprehensive 6-panel analysis...")
fig = ss.plot_comprehensive_analysis(all_stats, figsize=(18, 10))
plt.show()

# Save outputs
outdir = os.path.join(SIM_DIR, "spike_synchrony")
os.makedirs(outdir, exist_ok=True)

csv_path = os.path.join(outdir, "spike_sync_summary.csv")
all_stats.to_csv(csv_path, index=False)
print(f"\nSaved: {csv_path}")

fig_path = os.path.join(outdir, "sync_comprehensive_analysis.png")
fig.savefig(fig_path, dpi=150, bbox_inches="tight")
print(f"Saved: {fig_path}")

# Display summary statistics
print("\n=== Summary Statistics by Level ===")
for level in all_stats['level'].unique():
    df_level = all_stats[all_stats['level'] == level]
    print(f"\n{level}:")
    print(f"  Count: {len(df_level)}")
    print(f"  Mean sync: {df_level['mean_sync'].mean():.4f} ± {df_level['mean_sync'].std():.4f}")
    print(f"  Range: [{df_level['mean_sync'].min():.4f}, {df_level['mean_sync'].max():.4f}]")

all_stats.head(20)
```
